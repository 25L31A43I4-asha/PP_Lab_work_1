```python
# Experiment 1: Find the largest of three numbers

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: ")) 
num3 = float(input("Enter third number: "))

if (num1 >= num2) and (num1 >= num3):
   largest = num1
elif (num2 >= num1) and (num2 >= num3):
   largest = num2
else:
   largest = num3

print("The largest number is", largest)
```

```python
# Experiment 2: Display all prime numbers within an interval

lower = int(input("Enter lower limit: "))
upper = int(input("Enter upper limit: "))

print(f"\nPrime numbers between {lower} and {upper} are:")

for num in range(lower, upper + 1):
   if num <= 1:
       continue

   is_prime = True
   
   for i in range(2, int(num**0.5) + 1):
     if num % i == 0:
       is_prime = False
       break
   
   if is_prime:
        print(num,end=" ")
```

```python
# Experiment 3: Swap two numbers without using a temp variable
# Method 1: Python multiple assignment

a = int(input("Enter first number (a): "))
b = int(input("Enter second number (b): "))

print("\nBefore swapping: a =", a, "b =", b)
a, b = b, a

print("After swapping: a =", a, "b =", b)
```

```python
# Experiment 3: Swap two numbers without using a temp variable
# Method 2: Using arithmetic operations (no temp)

a = int(input("Enter first number (a): "))
b = int(input("Enter second number (b): "))

print("\nBefore swapping: a =", a, "b =", b)

a = a + b
b = a - b
a = a - b

print("After swapping: a =", a, "b =", b)
```

```python
# Experiment 4 : Demonstrate all operators in Python

print("="*50)
print("i) ARITHMETIC OPERATORS")
print("="*50)
x,y = 17, 5
print(f"x={x}, y={y}")
print("Addition  x + y =", x + y)
print("Subtraction  x - y =", x - y)
print("Multiplication  x * y =", x * y)
print("Division  x / y =", x / y)
print("Floor Division  x // y =", x // y)
print("Modulus  x % y =", x % y)
print("Exponent  x ** y =", x ** y)   

print("="*50)
print("ii) RELATIONAL (COMPARISON) OPERATORS")
print("="*50)
print("x==y:", x==y)
print("x!=y:", x!=y)
print("x>y:", x>y)
print("x<y:", x<y)
print("x>=y:", x>=y)
print("x<=y:", x<=y)

print("\n"+"="*50)
print("iii) ASSIGNMENT OPERATORS")
print("="*50)
a = 10
print("a =", a)
a += 5
print("a += 5 => a =", a)
a -= 3
print("a -= 3 => a =", a)
a *= 2
print("a *= 2 => a =", a)
a //= 4
print("a //= 4 => a =", a)

print("\n"+"="*50)
print("iv) LOGICAL OPERATORS")
print("="*50)
p, q = True, False
print("p =", p, "q =", q)
print("p and q:", p and q)
print("p or q:", p or q)
print("not p:", not p)

print("\n"+"="*50)
print("v) BITWISE OPERATORS")
print("="*50)
m,n = 10,4
print("m=",m,"(binary 1010),n=",n,"(binary 0100)")
print("m & n (AND) =", m & n)
print("m | n (OR) =", m | n)
print("m ^ n (XOR) =", m ^ n)
print("~m (NOT) =", ~m)
print("m << 1 (Left Shift) =", m << 1)
print("m >> 1 (Right Shift) =", m >> 1)

print("\n"+"="*50)
print("vi) TERNARY OPERATOR(Conditional Expression)")
print("="*50)
age = 20
status = "Adult" if age >= 18 else "Minor"
print(f"Age: {age}=> Status: {status}")

print("\n"+"="*50)
print("vii) MEMBERSHIP OPERATORS")
print("="*50)
fruits = ["apple", "mango", "banana"]
print("fruits =", fruits)
print("'mango' in fruits:", "mango" in fruits)
print("'grape' not in fruits:", "grape" not in fruits)

print("\n"+"="*50)
print("viii) IDENTITY OPERATORS")
print("="*50)
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1
print("list1 =", list1)
print("list2 =", list2)
print("list3 =", list3)
print("list1 == list2(same content):", list1 == list2)
print("list1 is list2(same object):", list1 is list2)
print("list1 is list3(same object):", list1 is list3)
```

```python
# Experiment 5: Add and multiply complex numbers

c1 = complex(input("Enter first complex number (in the form a+bj): "))
c2 = complex(input("Enter second complex number (in the form a+bj): "))

sum_result = c1 + c2
prod_result = c1 * c2

print("\n"+"="*45)
print("First complex number:", c1)
print("Second complex number:", c2)
print("-"*45)
print("Addition  c1 + c2 =", sum_result)
print("Multiplication  c1 * c2 =", prod_result)
print("="*45)

print("\nReal part of sum:", sum_result.real)
print("Imaginary part of sum:", sum_result.imag)
```

```python
# Experiment 6: Print multiplication table 

num = int(input("Enter a number to print its multiplication table: "))
print("\n"+"="*30)
print(f" Multiplication Table of {num}")
print("="*30)

for i in range(1, 11):
    print(f"{num} x {i:2} = {num * i:3}")

print("="*30)
```
