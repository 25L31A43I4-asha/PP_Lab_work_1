```python
# Exp 1: Function with multiple return values
def calculate(a,b):    
    addition=a+b    
    subtraction=a-b    
    multiplication=a*b    
    return addition, subtraction, multiplication 
num1=int(input("Enter first number: ")) 
num2=int(input("Enter second number: ")) 
add,sub,mul=calculate(num1,num2) 
print("Addition:",add) 
print("Subtraction:",sub) 
print("Multiplication:",mul)
```

```python
# Exp 2: Function with default arguments
def student_details(name, department="Artificial Intelligence"):    
    print("Name:", name)    
    print("Department:", department)    
student_details("Asha", "Artificial Intelligence")
student_details("Latha")
```

```python
# Exp 3: Find length of string without using len() 
text = input("Enter a string: ")
count = 0
for char in text:
    count += 1

print("Length:", count)    
```

```python
# Exp 4: Check Substring
main_string = input("Enter the main string: ")
substring = input("Enter the substring: ")

if substring in main_string:
    print("Substring found")
else:
    print("Substring not found")
```

```python
# Exp 5:List Addition, Insertion and Slicing
numbers = [10, 20, 30, 40, 50]

print("Original:", numbers)

numbers.append(60)
print("After Addition:", numbers)

numbers.insert(2, 25)
print("After Insertion:", numbers)

print("Slicing:", numbers[1:5])
```

```python
# Exp 6: Built-in Functions on List
marks = [85, 90, 78, 92, 88]

print("length:", len(marks))
print("Maximum:", max(marks))
print("Minimum:", min(marks))
print("Sum:", sum(marks))
print("Sorted:", sorted(marks))
```
